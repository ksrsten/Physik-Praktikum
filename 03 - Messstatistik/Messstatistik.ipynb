## Die Schätzung des "wahren Wertes"

Unsere Messdaten streuen also (hoffentlich) um den "wahren Wert". Die einfachste Methode diesen
abzuschätzen ist der Durchschnitt, bzw. das arithmetische Mittel:

$$\bar{x} = \cfrac{x_1+x_2+...+x_n}{n} = \cfrac{1}{n}\cdot\sum\limits_{i=1}^{n}x_i$$

Hierin sind $x_1$, $x_2$ usw. unsere einzelnen Messwerte, $n$ die Anzahl unserer Messwerte und $\bar{x}$ das arithmetische Mittel unserer Messwerte.

## Die Breite der Verteilung

Anscheinend liegen wir mit dem Durchschnitt in der Nähe von 200 g. Aber wir sehen im Diagramm auch, dass die Werte etwas streuen. Zur Quantifizierung dieser Streuung nutzt man die sogenannte Standardabweichung $s$ und die Varianz $Var$. Die Varianz ist definiert als Durchschnitt der quadratischen Abweichungen vom Mittelwert. Und die Standardabweichung ist deren Wurzel, wodurch wir wieder in der ursprünglichen Einheit landen. Mathematisch sieht das so aus:

$$Var = \cfrac{(x_1-\bar{x})^2+(x_2-\bar{x})^2+...+(x_n-\bar{x})^2}{n-1} = \cfrac{1}{n-1}\cdot\sum\limits_{i=1}^{n}(x_i-\bar{x})^2$$

$$s = \sqrt{Var}$$

Der folgende Code macht dies für uns. Beachte, dass dieser Code davon ausgeht, dass du den Durchschnitt der Werte zuvor mit "mean_weights" bezeichnet hast. Falls dies nicht so ist, kannst du den Code sicher schnell anpassen.

## Interpretation der Standardabweichung

Die Standardabweichung dient uns als Mass für die Streuung unserer Daten. Wir können grob drei Bereiche unterscheiden:

- Durchschnitt $\pm$ **eine** Standardabweichung (bei uns 194g - 208g): ein Grossteil unserer Daten liegt in diesem Bereich
- Durchschnitt $\pm$ **zwei** Standardabweichungen (bei uns 189g - 215g): die meisten unserer Daten liegen in diesem Bereich
- Durchschnitt $\pm$ **drei** Standardabweichungen (bei uns 183g - 221g): praktisch alle unserer Daten liegen in diesem Bereich

Würden wir noch mehr Messungen durchführen, so wird sich die Verteilung unserer Daten immer mehr der sogenannten Normalverteilung annähern. Für diese ist klar definiert, wie viele Werte innerhalb einer gewissen Anzahl Standardabweichungen um den Durchschnitt liegen (siehe Abbildung).

## Standardabweichung des arithmetischen Mittels

Bisher können wir die Streuung unserer Daten mathematisch (bzw. durch Python) bestimmen. Diese Streuung wird im Laufe vieler Messungen jedoch nicht merklich grösser oder kleiner werden, da sie eine Eigenschaft der Messung an sich darstellt. Wenn wir aber immer mehr messen, desto sicherer werden wir uns, dass unser gemessener Durchschnitt immer näher am "wahren Wert" liegen muss. Um nun diese geringer werdende Unsicherheit auszudrücken, verwendet man die sogenannte Standardabweichung des arithmetischen Mittels $s_n$. Diese wird berechnet durch:
$$s_n=\cfrac{s}{\sqrt{n}}$$
Nochmals in Bezug auf obige Abbildung liegt nun also der "wahre" Wert innerhalb von
$$x = x̄ ± s_n$$
mit einer Wahrscheinlichkeit von 68%. Da dies den meisten Menschen noch etwas zu unsicher ist, gibt man häufig eher einen Bereich von zwei Standardabweichungen an, was dann einer Sicherheit von 95% entspricht:
$$x = \bar{x} \pm 2\cdot s_n$$
