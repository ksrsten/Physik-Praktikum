## Einrichtung von Python

Um unsere Daten darstellen zu können, verwenden wir das Modul "matplotlib", zum Einlesen der Daten "pandas". Die folgenden Zeilen überprüfen, ob die Module schon installiert sind und kümmert sich darum, falls sie noch nicht installiert sind. Führe die folgenden Zeilen aus, indem du die Box anklickst und dann oben auf den Ausführen-Pfeil ▶ klickst. Diesen Code musst du nicht extra verstehen. Willst du die weiteren Code-Beispiele in einer lokalen Python-Installation ausführen (z.B. Thonny), so musst du dich selbst darum kümmern, dass die Module installiert sind.

```python
import subprocess

# Funktion zum Überprüfen, ob ein Modul bereits installiert ist
def check(package):
    ans = subprocess.run(["pip", "list"], capture_output=True, text=True)
    return package in ans.stdout

# Funktion, um ein Modul zu installieren
def install(package):
    subprocess.check_call(["pip", "install", package])

# Module, die installiert sein müssen
packages = ["matplotlib", "pandas"]

# Iteration durch die Modulliste
for package in packages:
    if not check(package):
        install(package)
```

## Laden der Module

Als Nächstes können wir die Module für die spätere Verwendung laden.

```python
# Modul zum Handling von Datendateien
import pandas as pd

# Modul zur Darstellung von Daten
import matplotlib.pyplot as plt
```

## Einlesen der Daten

Die Daten aus der Datei "sprint60m.txt" laden wir mithilfe der folgenden Zeilen und überprüfen in der Ausgabe, ob alles gut aussieht.

```python
# Daten aus Datei einlesen
data = pd.read_csv("sprint60m.txt")

# Daten zur Kontrolle ausgeben
print(data)

# Datenreihe in Liste umwandeln für spätere Verwendung
times = data["Zeiten in s"].to_list()
```

## Erzeugung des Diagramms

Die folgenden Zeilen erstellen uns das Diagramm. Das kannst du ruhig schon einmal ausprobieren. Wir können aber auch noch ein bisschen was einstellen. So kannst du z.B. mal die Breite der Balken oder deren Farbe ändern. Findest du, wo das möglich ist?

```python
# Diagrammvariablen erstellen
fig, ax = plt.subplots()

# Histogramm erstellen
bins = [9, 9.2, 9.4, 9.6, 9.8, 10, 10.2, 10.4, 10.6, 10.8, 11, 11.2] # Abschnitte für Balken definieren
ax.hist(times, bins, rwidth=0.9, color="teal") # Erstellung des Histogramms mit den Daten, den Abschnitten, der Balkenbreite und der Balkenfarbe

# Achsen passend mit Zahlen beschriften lassen
ax.set_xticks(bins)
ax.set_yticks([0,1,2,3])

# Achsen beschriften
ax.set_xlabel('Zeit in s')
ax.set_ylabel('Häufigkeit')

#Diagramm anzeigen
plt.show()
```

## Besserer Code

Vielleicht ist einigen von euch aufgestossen, dass wir die "bins" und die y-Achsen-Skalierung mit von Hand geschriebenen Listen erzeugt haben. Hätten wir andere Daten müssten wir diese immer wieder aufwändig neu anpassen. Stattdessen können wir versuchen, dies einwenig zu automatisieren. Warnung: Der folgende Code erfordert, dass man mitdenkt. Falls jemand nicht gerne nachdenkt, ist er hier vielleicht falsch.

```python
# Diagrammvariablen erstellen
fig, ax = plt.subplots()

# Histogramm erstellen
x_min = 9
x_max = 11.2
x_step = 0.2
x_bins = [x_min]
while x_bins[-1] < x_max:
    x_bins.append(bins[-1] + x_step)
ax.hist(times, bins, rwidth=0.9, color="teal") # Erstellung des Histogramms mit den Daten, den Abschnitten, der Balkenbreite und der Balkenfarbe

# Achsen passend mit Zahlen beschriften lassen
ax.set_xticks(bins)
y_min = 0
y_max = 3
y_step = 1
y_bins = [y_min]
while y_bins[-1] < y_max:
    y_bins.append(y_bins[-1] + y_step)
ax.set_yticks(y_bins)

# Achsen beschriften
ax.set_xlabel('Zeit in s')
ax.set_ylabel('Häufigkeit')

#Diagramm anzeigen
plt.show()
```
